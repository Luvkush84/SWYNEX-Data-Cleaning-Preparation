```python
import pandas as pd
import os

# Go one folder back and enter data folder
file_path = "../data/Online retail.xlsx"

# Check file
print("File Exists:", os.path.exists(file_path))

if os.path.exists(file_path):

    print("File Size:", os.path.getsize(file_path), "Bytes")

    # Load dataset
    df = pd.read_excel(file_path)

    print("\nDataset Loaded Successfully!")

    # Show first 5 rows
    display(df.head())

else:
    print("Excel file not found!")
```

```python
# Check total rows and columns

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python
# Check missing values in each column

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)

# Calculate total missing values

print("\nTotal Missing Values:", missing_values.sum())
```

```python
# Calculate missing values percentage

missing_count = df.isnull().sum()

missing_percentage = (missing_count / len(df)) * 100

# Create summary table

missing_report = pd.DataFrame({
    "Missing Values": missing_count,
    "Missing Percentage": missing_percentage
})

print(missing_report)
```

```python
# Check duplicate records
duplicate_count = df.duplicated().sum()

print("Total Duplicate Records:", duplicate_count)
```

```python
# Check column names and data types
print("Dataset Information:")
df.info()
```

```python
# Check CustomerID values before conversion

print("CustomerID Data Type:", df["CustomerID"].dtype)

print("\nFirst 10 CustomerID Values:")
print(df["CustomerID"].head(10))

print("\nMissing CustomerID:")
print(df["CustomerID"].isnull().sum())
```

```python
# Convert CustomerID from float to string
# Missing values will remain missing

df["CustomerID"] = df["CustomerID"].apply(
    lambda x: str(int(x)) if pd.notna(x) else pd.NA
)

# Check the result
print("New CustomerID Data Type:", df["CustomerID"].dtype)

print("\nFirst 10 CustomerID Values:")
print(df["CustomerID"].head(10))

print("\nMissing CustomerID:")
print(df["CustomerID"].isnull().sum())
```

```python
# Fill missing product descriptions

df["Description"] = df["Description"].fillna("Unknown Product")

# Verify the result

print("Missing Description Values:",
      df["Description"].isnull().sum())

print("\nUnknown Product Records:",
      (df["Description"] == "Unknown Product").sum())

print("\nSample Description Values:")
print(df["Description"].head(10))
```

```python
# Step 1: Count rows before removing duplicates
before_rows = len(df)

# Step 2: Remove duplicate records
df = df.drop_duplicates()

# Step 3: Count rows after removing duplicates
after_rows = len(df)

# Step 4: Calculate removed duplicates
removed_rows = before_rows - after_rows

# Display results
print("Rows Before Cleaning:", before_rows)
print("Rows After Cleaning:", after_rows)
print("Duplicate Rows Removed:", removed_rows)
print("Remaining Duplicate Rows:", df.duplicated().sum())
```

```python
# Remove extra spaces from text columns

df["Description"] = df["Description"].str.strip()
df["Country"] = df["Country"].str.strip()

# Verify text cleaning

print("Description spaces cleaned!")
print("Country spaces cleaned!")

print("\nSample Description:")
print(df["Description"].head(5).tolist())

print("\nCountry Names:")
print(df["Country"].unique()[:10])
```

```python
# Check Quantity values
print("QUANTITY ANALYSIS")

print("Negative Quantity:",
      (df["Quantity"] < 0).sum())

print("Zero Quantity:",
      (df["Quantity"] == 0).sum())

print("Minimum Quantity:",
      df["Quantity"].min())

print("Maximum Quantity:",
      df["Quantity"].max())


# Check UnitPrice values
print("\nUNIT PRICE ANALYSIS")

print("Zero UnitPrice:",
      (df["UnitPrice"] == 0).sum())

print("Negative UnitPrice:",
      (df["UnitPrice"] < 0).sum())

print("Minimum UnitPrice:",
      df["UnitPrice"].min())

print("Maximum UnitPrice:",
      df["UnitPrice"].max())
```

```python
# Display records with negative UnitPrice

negative_price = df[df["UnitPrice"] < 0]

print("Negative UnitPrice Records:")
display(negative_price)
```

```python
# Display records where UnitPrice is zero

zero_price = df[df["UnitPrice"] == 0]

print("Total Zero UnitPrice Records:", len(zero_price))

display(zero_price.head(10))
```

```python
# Count zero and negative UnitPrice records

print("Zero UnitPrice Records:", (df["UnitPrice"] == 0).sum())

print("Negative UnitPrice Records:", (df["UnitPrice"] < 0).sum())

print("Total Rows:", len(df))
```

```python
# FINAL DATASET VALIDATION

print("FINAL DATASET VALIDATION")

print("\nTotal Rows:", len(df))
print("Total Columns:", len(df.columns))

print("\nMissing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nCustomerID Data Type:", df["CustomerID"].dtype)
print("InvoiceDate Data Type:", df["InvoiceDate"].dtype)

print("\nDescription Missing:", df["Description"].isnull().sum())
```

```python
# Fix remaining missing Description

df["Description"] = df["Description"].fillna("Unknown Product")

# Verify cleaning

print("Missing Description:", df["Description"].isnull().sum())

print("Unknown Product Records:",
      (df["Description"] == "Unknown Product").sum())

print("Missing CustomerID:", df["CustomerID"].isnull().sum())

print("Duplicate Rows:", df.duplicated().sum())

print("Total Rows:", len(df))
```

```python

import os

# Create output folder if it does not exist
os.makedirs("../output", exist_ok=True)

# Save cleaned dataset
output_path = "../output/cleaned_online_retail.csv"

df.to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig"
)

# Confirm saved file
print("Dataset Saved Successfully!")

print("File Location:", os.path.abspath(output_path))
print("Total Rows:", len(df))
print("Total Columns:", len(df.columns))
print("File Size:", round(os.path.getsize(output_path) / (1024 * 1024), 2), "MB")
```
